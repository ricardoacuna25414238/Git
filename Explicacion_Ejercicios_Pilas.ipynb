## Ejercicio 1: Pila de números enteros

**Objetivo:** agregar números enteros a una pila y ofrecer un menú con 5 opciones: agregar, contar, mostrar, calcular promedio y salir.

**Diseño usado:** enfoque **funcional** — una lista global `pila` y una función independiente por cada opción del menú.

### Explicación de cada función

- **`agregar_numero()`**: pide un número por teclado dentro de un ciclo `while True` con manejo de errores (`try/except`), y lo agrega con `pila.append(numero)` (esto es el *push*).
- **`contar_elementos()`**: usa `len(pila)` para saber cuántos números hay.
- **`mostrar_elementos()`**: recorre la pila con `reversed(pila)` para mostrarla de la cima hacia la base, que es el orden lógico en que "saldrían" los elementos.
- **`calcular_promedio()`**: usa `sum(pila) / len(pila)`, cuidando primero que la pila no esté vacía (evita división entre cero).
- **`main()`**: contiene el ciclo del menú; según la opción elegida, llama a la función correspondiente.

A continuación, una versión de las funciones lista para ejecutar en esta celda (idéntica a la lógica del archivo `.py`), seguida de una **demostración simulada** sin usar `input()`, para que puedas ver el comportamiento sin necesidad de escribir datos.

```python
# Funciones del Ejercicio 1 (idénticas en lógica al archivo .py)

pila_ej1 = []

def agregar_numero(numero):
    """Agrega (push) un número entero a la pila."""
    pila_ej1.append(numero)
    print(f"El número {numero} fue agregado a la pila.")

def contar_elementos():
    print(f"La pila tiene {len(pila_ej1)} elemento(s).")

def mostrar_elementos():
    if not pila_ej1:
        print("La pila está vacía.")
        return
    print("Elementos de la pila (de la cima a la base):")
    for elemento in reversed(pila_ej1):
        print(f"  -> {elemento}")

def calcular_promedio():
    if not pila_ej1:
        print("La pila está vacía, no se puede calcular el promedio.")
        return
    promedio = sum(pila_ej1) / len(pila_ej1)
    print(f"El promedio de los números es: {promedio:.2f}")
```

### Demostración (sin `input()`, para ejecutar directo en el notebook)

Simulamos que un usuario agrega los números 10, 20 y 30, y luego consulta el menú.

```python
# Simulación de uso: agregar tres números y probar cada opción del menú
for n in [10, 20, 30]:
    agregar_numero(n)

contar_elementos()
mostrar_elementos()
calcular_promedio()
```

## Ejercicio 2: Pila de nombres

**Objetivo:** guardar el primer nombre de personas y ofrecer un menú de 8 opciones (más completo que el Ejercicio 1): agregar, eliminar, ver cima, buscar, contar, mostrar todos, limpiar y salir.

**Diseño usado:** también **funcional**, igual que el Ejercicio 1, pero con más operaciones típicas de una pila.

### Explicación de cada función

- **`agregar_nombre()`**: hace *push* de un nombre, validando que no esté vacío.
- **`eliminar_nombre()`**: hace *pop* — quita y muestra el nombre que estaba en la cima. Si la pila está vacía, avisa en lugar de fallar.
- **`mostrar_cima()`**: hace *peek* — muestra `pila[-1]` sin eliminarlo.
- **`buscar_nombre()`**: usa el operador `in` para verificar si el nombre existe en la pila, y calcula su posición desde la base.
- **`contar_elementos()`** y **`mostrar_elementos()`**: igual que en el Ejercicio 1.
- **`limpiar_pila()`**: usa `pila.clear()` para vaciarla por completo.

La diferencia clave con el Ejercicio 1 es que aquí la pila **crece y se reduce** dinámicamente (push y pop), mientras que en el Ejercicio 1 solo se agregaba.

```python
# Funciones del Ejercicio 2

pila_ej2 = []

def agregar_nombre(nombre):
    pila_ej2.append(nombre)
    print(f"'{nombre}' fue agregado a la pila.")

def eliminar_nombre():
    if not pila_ej2:
        print("La pila está vacía, no hay nada que eliminar.")
        return
    nombre = pila_ej2.pop()
    print(f"'{nombre}' fue eliminado de la pila.")

def mostrar_cima():
    if not pila_ej2:
        print("La pila está vacía.")
        return
    print(f"El elemento en la cima es: {pila_ej2[-1]}")

def buscar_nombre(nombre):
    if nombre in pila_ej2:
        posicion = len(pila_ej2) - pila_ej2[::-1].index(nombre) - 1
        print(f"'{nombre}' se encontró en la pila (posición {posicion} desde la base).")
    else:
        print(f"'{nombre}' no se encontró en la pila.")

def contar_elementos_ej2():
    print(f"La pila tiene {len(pila_ej2)} elemento(s).")

def mostrar_elementos_ej2():
    if not pila_ej2:
        print("La pila está vacía.")
        return
    print("Elementos de la pila (de la cima a la base):")
    for nombre in reversed(pila_ej2):
        print(f"  -> {nombre}")

def limpiar_pila():
    pila_ej2.clear()
    print("La pila ha sido limpiada.")
```

### Demostración

```python
# Simulación: agregar tres nombres, revisar la cima, buscar uno, eliminar y volver a mostrar
for nombre in ["Ana", "Luis", "Marta"]:
    agregar_nombre(nombre)

mostrar_cima()
buscar_nombre("Luis")
eliminar_nombre()
mostrar_elementos_ej2()
contar_elementos_ej2()
```

## Ejercicio 3: Pila de departamentos de Nicaragua (POO)

**Objetivo:** el mismo tipo de operaciones que el Ejercicio 2, pero implementadas con **Programación Orientada a Objetos**: una clase `PilaDepartamentos` donde cada operación es un **método**.

### ¿Por qué usar una clase aquí?

- **Encapsulamiento**: la lista interna `_elementos` está "protegida" dentro del objeto (el guion bajo `_` indica que es de uso interno). El resto del programa no manipula la lista directamente, sino a través de los métodos de la clase.
- **Cohesión**: todos los datos (la lista) y el comportamiento (agregar, eliminar, buscar, etc.) viven juntos dentro de un mismo objeto `PilaDepartamentos`.
- **Reutilización**: se podrían crear varias pilas independientes (`pila1 = PilaDepartamentos()`, `pila2 = PilaDepartamentos()`) sin que se mezclen sus datos, algo que sería más difícil con una sola lista global como en los Ejercicios 1 y 2.

### Explicación de los métodos

- **`__init__(self)`**: el *constructor*; se ejecuta al crear el objeto y crea la lista vacía `_elementos`.
- **`agregar(self, departamento)`**: hace *push*.
- **`eliminar(self)`**: hace *pop*, retornando `None` si la pila está vacía.
- **`ver_cima(self)`**: hace *peek*.
- **`buscar(self, departamento)`**: verifica si el departamento está en la pila usando `in`, y retorna `True`/`False`.
- **`contar(self)`**: retorna `len(self._elementos)`.
- **`imprimir(self)`**: recorre la pila con `reversed()`, igual que en los ejercicios anteriores.
- **`limpiar(self)`**: usa `clear()` sobre la lista interna.
- **`esta_vacia(self)`**: método auxiliar reutilizado por varios de los métodos anteriores para evitar errores al operar sobre una pila vacía.

Fíjate que la **lógica interna** (push, pop, peek, `reversed()`, `in`, `clear()`) es la misma que en los Ejercicios 1 y 2 — lo que cambia es la **organización del código**: en vez de funciones sueltas con una lista global, todo queda encapsulado dentro de un objeto.

```python
# Clase del Ejercicio 3

class PilaDepartamentos:
    def __init__(self):
        self._elementos = []

    def agregar(self, departamento):
        self._elementos.append(departamento)
        print(f"'{departamento}' fue agregado a la pila.")

    def eliminar(self):
        if self.esta_vacia():
            print("La pila está vacía, no hay nada que eliminar.")
            return None
        departamento = self._elementos.pop()
        print(f"'{departamento}' fue eliminado de la pila.")
        return departamento

    def ver_cima(self):
        if self.esta_vacia():
            print("La pila está vacía.")
            return None
        print(f"El elemento en la cima es: {self._elementos[-1]}")
        return self._elementos[-1]

    def buscar(self, departamento):
        if departamento in self._elementos:
            print(f"'{departamento}' se encuentra en la pila.")
            return True
        print(f"'{departamento}' no se encuentra en la pila.")
        return False

    def contar(self):
        print(f"La pila tiene {len(self._elementos)} elemento(s).")
        return len(self._elementos)

    def imprimir(self):
        if self.esta_vacia():
            print("La pila está vacía.")
            return
        print("Departamentos en la pila (de la cima a la base):")
        for departamento in reversed(self._elementos):
            print(f"  -> {departamento}")

    def limpiar(self):
        self._elementos.clear()
        print("La pila ha sido limpiada.")

    def esta_vacia(self):
        return len(self._elementos) == 0
```

### Demostración

```python
# Creamos un objeto de la clase y probamos sus métodos
pila_depto = PilaDepartamentos()

for depto in ["Managua", "Masaya", "Granada", "Estelí"]:
    pila_depto.agregar(depto)

pila_depto.ver_cima()
pila_depto.buscar("Masaya")
pila_depto.eliminar()
pila_depto.imprimir()
pila_depto.contar()
```
